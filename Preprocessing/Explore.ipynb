```python
#Understanding the Dataset


"""
Your dataset has 37 columns,
 organized naturally into several families. 
 The source describes one row as a (0.1° cell, day) observation,
 across 14 fire seasons from 2012–2025.


the format of my features is :    << #***** >>

"""
```

```python
"""
Spatial and temporal information

These columns tell us where and when the observation was made.


#cell_id:
A unique identifier for a 0.1° spatial cell.
It identifies the geographic grid cell, not a particular observation.
for example : cell_id = 340_60 can belong to many rows 
The source (kaggle) describes the dataset grain as one (cell_id, date) pair.

#lat:
Latitude of the cell centre.
It tells us the north/south geographic position of the cell.


#lon:
Longitude of the cell centre.
It tells us the east/west position of the cell.
7


so:
    lat = 34.0
    lon = 6.0
identifies approximately where the cell is located.

****************Important: lat and lon are geographic coordinates,
 not the weather measurements themselves. 
 The dataset documentation says they are intended for mapping 
 and should not be used as normal model features
because they can introduce a geographic shortcut.**************

#date: year-month-day 

#year and #doy : can be calculated from date 
and i see them as irrelevant to the training. (doy = day of year from 1 to 365/366)



"""
```

```python
"""
Target and fire-detection information :

#label:v(int64)
label = 1 → qualifying fire detection
label = 0 → no qualifying fire detection

its not  certain for label = 0

note : The source explicitly warns about this limitation
 because satellite detection 
can be affected by things such as cloud.

#sample_kind:
{positive,matched_negative}

A positive row belongs to the positive fire examples.

A matched_negative row is a negative example deliberately
selected according to the dataset's negative-sampling procedure.
Specifically, negatives are matched to positive observations
in the same cell and season under the stated temporal constraints.


#n_detections:
The number of qualifying VIIRS
(VIIRS is the satellite sensor used to detect the fires in your dataset.)
detection pixels
associated with that cell-day.
A negative has: : n_detections = 0.

#max_frp:
Maximum Fire Radiative Power (FRP) associated with the fire
detections for that cell-day.
the dataset has NaN for max_frp on all negative samples, 
which makes it another perfect label leak.



this features are dangerous cause they can leak infos into our model .
we should drop them before training.



"""
```

```python
"""
Same-day weather

These describe the meteorological conditions for that particular day.
The dataset contains nine such variables.


#temp_max:
Maximum temperature during the day.

#temp_min:
Minimum temperature during the day.


#temp_mean:
Mean temperature during the day.


#rh_min:
Minimum relative humidity during the day.
Relative humidity describes how much water vapour
is present in the air relative to the amount the air 
could hold at that temperature.

#rh_mean:
Mean relative humidity during the day.

#wind_max:
Maximum wind speed during the day.
It tells us the strongest wind condition reached during that day.

#wind_mean:
Mean wind speed during the day.

#precip_sum:
Total precipitation during the day.
how much water has fallen during that day.

#vpd_max
Maximum Vapour Pressure Deficit (VPD) during the day.
VPD measures the atmospheric drying demand: roughly, how strongly the atmosphere is able to draw moisture away from surfaces and vegetation.













"""
```

```python
"""
Antecedent dryness / recent weather history

These variables don't describe only today. They contain information about what happened before today.
the source explicitly says they are backward-looking.


#precip_7d:

Total precipitation accumulated over the previous 7 days.


#precip_30d:

Total precipitation accumulated over the previous 30 days.

#days_since_rain_1mm:

Number of days since the last day 
on which precipitation reached at least 1 mm.


#temp_max_7d_mean:

Mean of the daily maximum temperature over the previous 7 days.


#rh_min_7d_mean:

Mean of the daily minimum relative humidity over the previous 7 days.







"""
```

```python
"""

Canadian Fire Weather Index variables

These six variables are derived indices 
rather than direct weather measurements.



#ffmc:

Fine Fuel Moisture Code

An FWI component representing 
the moisture/dryness state of fine fuels.


#dmc:

Duff Moisture Code

An FWI component related to the moisture 
condition of the duff/organic material layer.

#dc:

Drought Code

Represents a longer-term dryness/drought component.

So compared with something like today's precipitation, 
dc carries more accumulated historical information.


#isi:

Initial Spread Index

An index associated with the potential for fire spread, 
incorporating the relevant weather/fuel dryness effects
in the FWI system.


#bui:

Buildup Index

A combined index representing the buildup of available 
dry fuel represented by the relevant FWI components.

#fwi:

Fire Weather Index

The overall FWI-system fire-weather indicator
derived from the system's components.

note :


The source explicitly warns that these FWI values
are much higher than canonical published ranges because
of the particular implementation used here. 
They are intended to be internally consistent for this dataset.







"""
```

```python
"""
Fire history

#days_since_last_fire:

This describes the previous fire history of the cell.

days_since_last_fire = 9999 means no previous record is available.




"""
```

```python
"""
Vegetation variables

These come from MODIS vegetation data.
The source says the vegetation measurements are based 
on MOD13Q1 v061 16-day composites aggregated over
the whole 0.1° cell.

#ndvi:

Normalized Difference Vegetation Index (NDVI) for the 
current vegetation observation.

In simple terms, it measures vegetation greenness/condition 
from the satellite-derived vegetation product.


#ndvi_normal:

The historical normal NDVI 
for that cell and corresponding seasonal period.



#ndvi_anomaly:

The difference between the current NDVI and its normal.
positive anomaly → current NDVI above normal
negative anomaly → current NDVI below normal


#ndvi_change_32d:

Change in NDVI over approximately the previous 32 days.


#ndvi_stale_days:

How many days old the NDVI composite being used for the row is.

This exists because MODIS provides 16-day composites 
rather than a fresh vegetation measurement for every single day.
The dataset deliberately uses a composite only after its
observation window has 
closed before the label date.

So:

ndvi_stale_days = 2

means the NDVI measurement being used is 2
 days old relative to the row's date.

It does not mean the vegetation itself is "stale."

It describes the age of the available satellite observation.


#split:

This is machine-learning metadata.

train → 2012–2023
val   → 2024
test  → 2025

The source explicitly defines the split this way.

So it doesn't describe wildfire conditions.

It tells us:

Which temporal dataset partition should this row belong to?

"""
```

```python
#understand the target
"""

| Feature                | What it tells us                                     | Leakage?                                 |
| ---------------------- | ---------------------------------------------------- | ---------------------------------------- |
| `label`                | The target itself                                    | —                                        |
| `sample_kind`          | Whether the row was constructed as positive/negative | **Yes**                                  |
| `n_detections`         | Number of VIIRS fire detections for that cell-day    | **Yes**                                  |
| `max_frp`              | Fire Radiative Power from those detections           | **Yes**                                  |
| `days_since_last_fire` | Whether/how recently a **previous** fire occurred    | **No, if computed only from past dates** |

so before training we need to drop them !!!!!!!


"""



```
